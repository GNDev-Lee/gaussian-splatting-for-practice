## Find Given Value From an Array

```python
arr = np.array([[1, 2, 3],[1, 1, 1]])
print(arr)
print(arr == 1)
print(np.where(arr == 1))
wh = np.where(arr == 1)
for i in range(len(wh[0])):
    assert arr[wh[0][i], wh[1][i]] == 1
```

```python
arr = np.array([1, 2, 3])
print(arr)
print(arr == 1)
print(np.where(arr == 1))
wh = np.where(arr == 1)
for i in range(len(wh[0])):
    assert arr[wh[0][i]] == 1
```

# Imports

```python
import vispy
print(vispy.sys_info()) # Shows what modules we are using to render.
```

```python
import numpy as np
import vispy.app
import vispy.scene
from vispy.scene import visuals

# Setting Vispy backend as 'jupyter_rfb' explicitly
vispy.app.use_app('jupyter_rfb')

# Make a canvas and add simple view
canvas = vispy.scene.SceneCanvas(keys='interactive', show=True)
view = canvas.central_widget.add_view()

# generate data
pos = np.random.normal(size=(10000, 3), scale=0.2)
centers = np.random.normal(size=(50, 3))
indexes = np.random.normal(size=10000, loc=centers.shape[0] / 2,
                           scale=centers.shape[0] / 3)
indexes = np.clip(indexes, 0, centers.shape[0] - 1).astype(int)
symbols = np.random.choice(['o', '^'], len(pos))
scales = 10**(np.linspace(-2, 0.5, centers.shape[0]))[indexes][:, np.newaxis]
pos *= scales
pos += centers[indexes]

# create scatter object and fill in the data
scatter = visuals.Markers()
scatter.set_data(pos, edge_width=0, face_color=(1, 1, 1, .5), size=5, symbol=symbols)

view.add(scatter)

view.camera = 'turntable'  # or try 'arcball'

# add a colored 3D axis for orientation
axis = visuals.XYZAxis(parent=view.scene)

# call canvas object to render
canvas
```

# Using Open3D, Vispy to read ply files

```python
import open3d as o3d
from vispy import app, scene
```

```python
app.use_app('jupyter_rfb') # It allows for direct rendering within the Jupyter Notebook without opening a separate window.
```

```python
# 1. Generating Canvas and View
canvas = scene.SceneCanvas(keys='interactive', show=True, bgcolor = 'white', size = (800,600))
view = canvas.central_widget.add_view()
```

```python
# 2. Load PLY file using Open3D(Point Cloud)
file_path = "./GaussianTest/Test/sparse/0/points3D.ply" # You can modify file_path
pcd = o3d.io.read_point_cloud(file_path)
vertices = np.asarray(pcd.points)
```

```python
# 3. Visualize points with Markers objects
scatter1 = scene.visuals.Markers()
# You can change points' size by modifying size parameter
scatter1.set_data(vertices,face_color=(1.0,1.0,1.0,1.0), edge_color=None, size=4) # face_color(R,G,B, Opacity)
```

```python
view.add(scatter1) # Adding scatter1 on view
```

```python
# 5. Set camera
view.camera = 'arcball' # else 'turntable'
view.camera.set_range(margin=0)
view.camera.scale_factor *= 0.6
```

```python
# 6. Render by calling scene's object canvas
canvas
```
